```python
from PIL import Image
input_path='/content/poster_rev5 (1).png'
out_path='output_300dpi.tif'
im=Image.open(input_path)

# Desired output dimensions in meters and DPI
desired_width_m = 3.0
desired_height_m = 2.25
dpi = 300

# Convert meters to inches (1 meter = 39.3701 inches)
width_inches = desired_width_m * 39.3701
height_inches = desired_height_m * 39.3701

# Calculate target pixel dimensions (width, height)
target_width_pixels = int(width_inches * dpi)
target_height_pixels = int(height_inches * dpi)

# Resize the image to the calculated pixel dimensions
im = im.resize((target_width_pixels, target_height_pixels), Image.LANCZOS)

im.save(out_path, dpi=(dpi,dpi), compression='tiff_deflate')
out_path
```
